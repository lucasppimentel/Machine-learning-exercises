# Modelagem de K-Nearest Neighbors

```python
def predict(X_train, y_train, X_test, k=5):
    y_pred = []
    
    for test_point in X_test:
        # Distância Euclidiana (Soma dos quadrados por coluna ao longo do axis=1)
        distances = np.sum((X_train - test_point) ** 2, axis=1)
        
        # Obtém os índices dos K menores valores usando argsort
        k_indices = np.argsort(distances)[:k]
        
        # Obtém os rótulos dos K vizinhos mais próximos
        k_nearest_labels = y_train[k_indices]
        
        # Votação por Maioria (Moda)
        # Bincount conta a frequência dos inteiros e argmax seleciona a classe mais frequente
        most_common = np.bincount(k_nearest_labels).argmax()
        y_pred.append(most_common)
        
    return np.array(y_pred)
```

```python
# Testando o algoritmo
k = 5
predictions = predict(X_train, y_train, X_test, k=k)

# Avaliação da Acurácia
accuracy = np.mean(predictions == y_test)
print(f"Acurácia do KNN (K={k}): {accuracy * 100}%")
```

# Pontos Extra

## 1. Ponderação por Distância (*Distance-Weighted KNN*)

No KNN tradicional (votação uniforme), todos os $K$ vizinhos possuem o mesmo peso na decisão final ($w_i = 1$). No entanto, se um vizinho está extremamente perto do ponto de teste e os outros $K-1$ estão distantes, dar o mesmo peso a todos pode gerar classificações errôneas.

Para resolver isso, atribui-se a cada vizinho um **peso inversamente proporcional à sua distância** em relação à amostra de teste:

$$w_i = \frac{1}{d(x_{teste}, x_i) + \epsilon}$$

* $d(x_{teste}, x_i)$: Distância entre a amostra de teste e o $i$-ésimo vizinho.
* $\epsilon$: Uma constante infinitesimal (ex: $10^{-5}$) para evitar divisão por zero se a distância for nula.

---

## 2. A Maldição da Dimensionalidade (*Curse of Dimensionality*)

O KNN baseia-se fundamentalmente na suposição de proximidade geométrica. Contudo, conforme o número de características (dimensões $D$) aumenta, a densidade de amostras no espaço de entrada cai exponencialmente.

### Principais Impactos no KNN:
1. **Perda de Significado da Distância:** Em espaços de altíssima dimensão, a diferença entre a distância do vizinho "mais próximo" e do "mais distante" tende a zero:
   $$\lim_{D \to \infty} \frac{d_{max} - d_{min}}{d_{min}} \to 0$$
   Todos os pontos ficam praticamente à mesma distância uns dos outros.
2. **Necessidade Exponencial de Dados:** Para manter a mesma densidade de amostragem no espaço, a quantidade de dados necessária cresce na ordem de $\mathcal{O}(2^D)$.
3. **Sensibilidade ao Ruído:** Recursos irrelevantes adicionam ruído à distância euclidiana, distorcendo os vizinhos verdadeiros.

> **Soluções Práticas:** Aplicação de redução de dimensionalidade (ex: **PCA**) e seleção criteriosa de características (*Feature Selection*) antes de rodar o KNN.

---

## 3. Estruturas de Dados Otimizadas: Kd-Trees

A implementação trivial do KNN (busca por força bruta) compara o ponto de teste contra todas as $N$ amostras do treino. Isso resulta em uma complexidade de tempo de inferência de **$\mathcal{O}(N \cdot D)$ por predição**, tornando o algoritmo inviável para grandes datasets.

Uma **Kd-Tree (K-dimensional Tree)** é uma estrutura de dados de árvore de busca binária que particiona recursivamente o espaço de recursos em hiperplanos ortogonais aos eixos.

### Como Funciona a Construção:
1. Seleciona-se um eixo/dimensão $d$ (alternando a cada nível da árvore: $d = \text{nível} \pmod D$).
2. Encontra-se a **mediana** dos dados ao longo dessa dimensão.
3. Divide-se o espaço em dois subespaços: amostras com valor $< \text{mediana}$ vão para o filho esquerdo; amostras $\ge \text{mediana}$ vão para o filho direito.
4. Repete-se o processo recursivamente.

### Comparativo de Complexidade de Tempo:

| Método | Construção da Estrutura | Inferência por Ponto (Casos Médios) | Inferência (Pior Caso / Altas Dimensões) |
| :--- | :--- | :--- | :--- |
| **Força Bruta** | $\mathcal{O}(1)$ | $\mathcal{O}(N \cdot D)$ | $\mathcal{O}(N \cdot D)$ |
| **Kd-Tree** | $\mathcal{O}(D \cdot N \log N)$ | $\mathcal{O}(D \log N)$ | $\mathcal{O}(N \cdot D)$ |

*Nota: Quando a dimensão $D$ é muito alta ($D \gtrsim 20$), a eficiência da Kd-Tree se deteriora e a busca se aproxima do desempenho da força bruta ($\mathcal{O}(N)$). Nesses cenários, costuma-se recorrer à **Ball Tree** ou algoritmos de vizinhos mais próximos aproximados (ANN - Approximate Nearest Neighbors).*
